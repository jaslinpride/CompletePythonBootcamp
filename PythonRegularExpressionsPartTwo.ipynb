```python
text = 'My phone number is 408-555-1234'
```

```python
import re
```

```python
phone = re.search('408-555-1234',text)
```

```python
phone
```

We may not always know the actual number, we may just know the pattern

Recall, when working with strings, back slashes indicate special escape characters, such as “\n” for “new line” or “\t” for “tab”. To tell Python that we are using these as a pattern for a regular expression add an “r” in front of the string.

```python
phone = re.search(r'\d\d\d-\d\d\d-\d\d\d\d',text)
```

```python
phone
```

```python
phone.group()
```

Looking for a pattern that included 20 or 100 digits

Use quantifiers to indicate repetition of the same character

```python
phone = re.search(r'\d{3}-\d{3}-\d{4}',text)
```

```python
phone
```

Performing two tasks:

1: Find Phone Numbers
2: Quickly extract area code (first 3 digits of phone number)

Use groups for any general tasks that involves grouping together regular expressions. This allows us to later break them down.  


Compile Function: compiles together different regular expression pattern codes

```python
phone_pattern = re.compile(r'(\d{3})-(\d{3})-(\d{4})')
```

```python
results = re.search(phone_pattern,text)
```

```python
results.group()
```

```python
results.group(1)
```

```python
results.group(2)
```

```python
results.group(3)
```
