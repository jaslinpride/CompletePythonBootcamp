Compressing Files to Send in an E-Mail

```python
import zipfile
```

```python
comp_file = zipfile.ZipFile('comp_file.zip','w')
```

Compress Text Files and Insert them into Zip File

```python
comp_file.write('fileone.txt',compress_type=zipfile.ZIP_DEFLATED)
```

```python
comp_file.write('filetwo.txt',compress_type=zipfile.ZIP_DEFLATED)
```

Close File

```python
comp_file.close()
```

Extract Items from Zip File

```python
zip_obj = zipfile.ZipFile('comp_file.zip','r')
```

Extract Method: Looking for 1 Specific File

```python
zip_obj.extract('extracted_content')
```

Extract All Method: Looking for Everything

```python
zip_obj.extractall('extracted_content')
```

Shell Utility Library

Give Output File Name

```python
output_filename = 'example'
```

Call Shell Utility Library

```python
shutil.make_archive(output_filename,'zip',dir_to_zip)
```
