"or" Operator

```python
import re
```

```python
re.search(r'cat','The cat is here')
```

"pipe" Operator

```python
re.search(r'cat|dog','The dog is here')
```

```python
re.search(r'cat|dog','The cat is here')
```

"Wildcard" Operator

```python
re.findall(r'at','The cat in the hat sat there.')
```

Grabbing the letter in front of "at"

```python
re.findall(r'.at','The cat in the hat sat there.')
```

```python
re.findall(r'...at','The cat in the hat went splat.')
```

Starts with "^"

```python
re.findall(r'^\d','1 is a number')
```

Only searches if the entire text starts with a number:

```python
re.findall(r'^\d','The 2 is a number')
```

Ends with "$"

```python
re.findall(r'\d$','The number is 2')
```

Exclusion

```python
phrase = 'there are 3 numbers 34 inside 5 this sentence'
```

Get back everything that is not a number in this sentence:

```python
pattern = r'[^\d]'
```

```python
re.findall(pattern,phrase)
```

```python
pattern = r'[^\d]+'
```

```python
re.findall(pattern,phrase)
```

exclusion is used to commonly get rid of punctuation in sentences

```python
test_phrase = 'This is a string! But it has punctuation. How can we remove it?'
```

```python
re.findall(r'[^!.?]+',test_phrase)
```

Adding a space after the question mark

```python
re.findall(r'[^!.? ]+',test_phrase)
```

```python
clean = re.findall(r'[^!.?]+',test_phrase)
```

```python
''.join(clean)
```

Inclusion

Square Brackets: for grouping things together

```python
text = 'Only find the hyphen-words in this sentence. But you do not know how long-ish they are'
```


Find words that have a hyphen or dash in the middle of them:


```python
pattern = r'[\w]+'
```

```python
re.findall(pattern,text)
```

```python
pattern = r'[\w]+-[\w]+'
```

```python
re.findall(pattern,text)
```

Removing braces gets the same result

```python
pattern = r'\w+'
```

```python
re.findall(pattern,text)
```

```python
pattern = r'\w+-\w+'
```

```python
re.findall(pattern,text)
```

Parentheses

Used for multiple options

```python
text = 'Hello, would you like some catfish?'
texttwo = "Hello, would you like to take a catnap?"
textthree = "Hello, have you seen this caterpillar?"
```

```python
re.search(r'cat(fish|nap|claw)',text)
```

```python
re.search(r'cat(fish|nap|claw)',texttwo)
```

```python
re.search(r'cat(fish|nap|claw)',textthree)
```

blank (does not list “|erpillar”)

```python
re.search(r'cat(fish|nap|erpillar)',textthree)
```
