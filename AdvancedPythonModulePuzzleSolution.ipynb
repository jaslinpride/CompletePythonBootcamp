**Use the Python os module and regular expressions to iterate through all the text files, open it, search for the telephone number**

**Create a regular expression to find the link **

```python
import re
```

```python
pattern = r'\d{3}-\d{3}-\d{4}'
```

```python
test_string = 'here is a phone number 123-123-1234'
```

**Make sure the pattern can find the phone number:**

```python
re.findall(pattern,test_string)
```

**Make it into a function:**

The function is going to accept a file (a text file), open it up and then search for the pattern within that file.

```python
def search(file,pattern=r'\d{3}-\d{3}-\d{4}'):
  f = open(file,'r')
  text = f.read()

  if re.search(pattern,text):
    return re.search(pattern,text)
  else:
    return ''
```

Walk through all the files using the os module in folders “One,” “Two,” “Three,” “Four,” and “Five.”

```python
import os
```

```python
results = []
```

```python
os.getcwd()+'\\extracted_content'
```

```python
for folder,sub_folders,files in os.walk(os.getcwd()+'\\extracted_content'):

  for f in files:
    full_path = folder+'\\'+f

    results.append(search(full_path))
```

**What are the results looking like? **

```python
for r in results:
  if r != '':
    print(r.group())
```
