Map Function

```python
def square(num):
  return num**2
```

```python
my_nums = [1,2,3,4,5]
```


Use a for loop or the map function to apply the square function to every single number in the my_nums list


```python
for item in map(square,my_nums):
  print(item)
```

Another Method: Built-in List Function

```python
list(map(square,my_nums))
```

Example: Function that works with strings

```python
def splicer(mystring):
  if len(mystring)%2 == 0:
    return 'EVEN'
  else:
    return mystring[0]
```

```python
names = ['Andy','Eve','Sally']
```

```python
list(map(splicer,names))
```

Filter Function: Returns an iterator yielding those items of the iterable for which when you pass in the item into the function, it is true. Meaning, you need to filter by a function that returns either True or False.

```python
def check_even(num):
  return num%2 == 0
```

```python
mynums = [1,2,3,4,5,6]
```

Only want to grab even numbers from this list.

Transform it, or

```python
list(filter(check_even,mynums))
```

Iterate through the list

```python
for n in filter(check_even,mynums):
  print(n)
```

Lambda Expressions

```python
def square(num):
  result = num ** 2
  return result
```

```python
square(3)
```

Converting a function step-by-step to a Lambda Expression

```python
def square(num):
  return num ** 2
```

```python
square(3)
```

Can also write on one line

```python
def square(num): return num ** 2
```

Use the keyword “lambda”

```python
lambda num: num ** 2
```

Could assign it to square

```python
square = lambda num: num ** 2
```

```python
square(3)
```

```python
square(5)
```

Use Lambda Expressions in conjunction with other functions, such as map and filter

Lambda with Map Function

```python
list(map(lambda num:num**2,mynums))
```

Lambda with Filter Function

```python
list(filter(lambda num:num%2 == 0,mynums))
```
