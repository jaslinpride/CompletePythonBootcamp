Enumerate Function

```python
index_count = 0

for letter in 'abcde':
  print('At index {} the letter is {}'.format(index_count,letter))
  index_count += 1
```

This is a common function. Use enumerate to replicate this.

```python
index_count = 0
word = 'abcde'

for letter in word:

  print(word[index_count])
  index_count += 1
```

Python built in the "enumerate" function

```python
word = 'abcde'

for item in enumerate(word):
  print(item)
```

```python
word = 'abcde'

for index,letter in enumerate(word):
  print(index)
  print(letter)
  print('\n')
```

Zip Function: zips together two lists, opposite of enumerate

```python
mylist1 = [1,2,3]
mylist2 = ['a','b','c']
```

Zip the two lists together using the zip function

```python
for item in zip(mylist1,mylist2):
  print(item)
```

```python
mylist1 = [1,2,3]
mylist2 = ['a','b','c']
mylist3 = [100,200,300]
```

```python
for item in zip(mylist1,mylist2,mylist3):
  print(item)
```

In Zip Functions, what happens if the lists are uneven?

```python
mylist1 = [1,2,3,4,5,6]
mylist2 = ['a','b','c']
mylist3 = [100,200,300]
```

```python
for item in zip(mylist1,mylist2,mylist3):
  print(item)
```

To get the list itself, just cast it using “list”


```python
list(zip(mylist1,mylist2))
```

Unpack using Tuple Unpacking

```python
for a,b,c in zip(mylist1,mylist2,mylist3):
  print(b)
```
