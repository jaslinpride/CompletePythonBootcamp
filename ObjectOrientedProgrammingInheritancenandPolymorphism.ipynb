**Inheritance**: a way to form new classes using classes that have already been defined

**Benefits**
*   Ability to reuse code
*   Reduce the complexity of a program





```python
class Animal():

  def __init__(self):
    print("ANIMAL CREATED")
```

```python
myanimal = Animal()
```

Adding more Methods

```python
class Animal():

  def __init__(self):
    print("ANIMAL CREATED")

  def who_am_i(self):
    print("I am an animal")

  def eat(self):
    print("I am eating")
```

```python
myanimal = Animal()
```

```python
myanimal.eat()
```

```python
myanimal.who_am_i()
```

Derived Class: deriving features from the base class of Animal  

```python
class Dog(Animal):

  def __init__(self):
    Animal.__init__(self)
    print("Dog Created")
```

```python
mydog = Dog()
```

The old methods, "who_am_i" and "eats," are now available for “mydog”

```python
mydog.eat()
```

What if I want to overwrite one of the older methods?

```python
class Dog(Animal):

  def __init__(self):
    Animal.__init__(self)
    print("Dog Created")

  def who_am_i(self):
    print("I am a dog!")
```

```python
mydog = Dog()
```

```python
mydog.eat()
```

```python
mydog.who_am_i()
```

Also able to add on methods

```python
class Dog(Animal):

  def __init__(self):
    Animal.__init__(self)
    print("Dog Created")

  def eat(self):
    print("I am a dog and eating")

  def bark(self):
    print("WOOF!")
```

```python
mydog = Dog()
```

```python
mydog.eat()
```

```python
mydog.who_am_i()
```

Polymorphism

```python
class Dog():

  def __init__(self,name):
    self.name = name

  def speak(self):
    return self.name + " says woof!"
```

```python
class Cat():

  def __init__(self,name):
    self.name = name

  def speak(self):
    return self.name + " says meow!"
```

```python
niko = Dog("niko")
felix = Cat("felix")
```

```python
print(niko.speak())
```

```python
print(felix.speak())
```

Polymorphism with a For Loop

```python
for pet in [niko,felix]:

  print(type(pet))
  print(pet.speak())
```

Polymorphism through a Function

```python
def pet_speak(pet):
  print(pet.speak())
```

```python
pet_speak(niko)
```

```python
pet_speak(felix)
```

Using Abstract Classes and Inheritance

```python
class Animal():

  def __init__(self,name):
    self.name = name

  def speak(self):
    raise NotImplementedError("Subclass must implement this abstract method")
```

Creating an instance of Animal (something we are not expected to do)

```python
myanimial = Animal('fred')
```

```python
myanimal.speak()
```

The reason this is an “abstract method” is because in the base class itself, it does not actually do anything. It is expecting you to inherit the “Animal” class and then overwrite the speak method.

```python
class Dog(Animal):

  def speak(self):
    return self.name + " says woof!"
```

```python
class Cat(Animal):

  def speak(self):
    return self.name + " says meow!"
```

```python
fido = Dog("Fido")
```

```python
isis = Cat("Isis")
```

```python
print(fido.speak())
```

```python
print(isis.speak())
```
