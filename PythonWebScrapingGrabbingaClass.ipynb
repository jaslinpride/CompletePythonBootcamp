Wikipedia Article: Grace Hopper
Grab all the strings in the Table of Contents


```python
import requests
```

```python
res = requests.get('https://en.wikipedia.org/wiki/Grace_Hopper')
```

Make a soup out of the text from the request

```python
import bs4
```

```python
soup = bs4.BeautifulSoup(res.text,"lxml")
```

```python
soup
```

```python
soup.select('.toctext')
```

Loop through information and grab the text attribute from the items

```python
soup.select('.toctext')[0]
```
