```python
mystring = 'hello'
```

Make a list of every character in this

```python
mylist = []

for letter in mystring:
  mylist.append(letter)
```

```python
mylist
```

More efficient method

```python
mylist = [letter for letter in mystring]
```

```python
mylist
```

```python
mylist = [x for x in 'word']
```

Range

```python
mylist = [x for x in range(0,11)]
```

```python
mylist
```

```python
mylist = [num for num in range(0,11)]
```

```python
mylist
```

Can perform operations on the first variable name

```python
mylist = [num**2 for num in range(0,11)]
```

```python
mylist
```

Can add if statement where you only grab even numbers

```python
mylist = [x for x in range(0,11) if x%2==0]
```

```python
mylist
```

Temperatures

```python
celsius = [0,10,20,34.5]

fahrenheit = [((9/5)*temp + 32) for temp in celsius]
```

```python
fahrenheit
```

The same thing as

```python
fahrenheit = []

for temp in celsius:
  fahrenheit.append(((9/5)*temp + 32))
```

```python
fahrenheit
```

"If else" statements in List Comprehension

```python
results = [x if x%2==0 else 'ODD' for x in range(0,11)]
```

```python
results
```

Nested Loops in List Comprehension

```python
mylist = []

for x in [2,4,6]:
  for y in [100,200,300]:
    mylist.append(x*y)
```

```python
mylist
```

Nested Loops in List Comprehension: To make the math more obvious

```python
mylist = []

for x in [2,4,6]:
  for y in [1,10,1000]:
    mylist.append(x*y)
```

```python
mylist
```

Nested Loops in List Comprehension: Consolidated

```python
mylist = [x*y for x in [2,4,6] for y in (1,10,1000)]
```

```python
mylist
```
