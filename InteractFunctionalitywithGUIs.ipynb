```python
from ipywidgets import interact,interactive,fixed
import ipywidgets as widgets
```

Passing in Integers

```python
def func(x):
  return x
```

```python
interact(func,x=10)
```

```python
def func(x):
  return x**2
```

```python
interact(func,x=10)
```

Passing in Booleans

```python
def func(x):
  return x
```

```python
interact(func,x=True)
```

Passing in a String

```python
interact(func,x='Hello')
```

Passing in Functions using a Decorator

```python
@interact(x=True,y=1.0)
def g(x,y):
  return(x,y)
```

Keeping 1 of the Values Fixed

```python
@interact(x=True,y=fixed(1.0))
def g(x,y):
  return(x,y)
```

```python
interact(func,x=fixed('Hello'))
```

Widget Abbreviations

Your minimum is defined by “x” and your maximum is defined by “x” times 3

```python
interact(func,x=1)
```

```python
interact(func,x=10)
```

Use Widgets to Change Minimum, Maximum, Step Size, and Value

```python
interact(func,x=widgets.IntSlider(min=-100,max=100,step=1,value=0))
```

Abbreviated

```python
interact(func,x=(-100,100,1))
```

Floating Point Abbreviation

```python
interact(func,x=(-10.0,10.0,0.1))
```

Interact Decorator with Floating Point Abbreviation

```python
@interact(x=(0.0,20.0,0.5))
def h(x=5.0):
  return x
```

Dropdown Menu Abbreviation

```python
interact(func,x='hello')
```

List of Strings Passed In

```python
interact(func,x=['hello','option 2','option 3'])
```

Passing in a Dictionary

```python
interact(func,x={'one':10,'two':20})
```

Interactive Function

```python
from IPython.display import display

def f(a,b):
  display(a+b)
  return a+b
```

```python
w = interactive(f,a=10,b=20)
```

```python
type(w)
```

```python
w.children
```
