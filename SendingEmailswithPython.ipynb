Sending E-Mails

Import the SMTP

```python
import smtplib
```

Create an SMTP Object for the Server

```python
smtp_object = smtplib.SMTP('smtp.gmail.com',587)
```

Run an EHLO Method Command to Greet the Server and Establish a Connection

```python
smtp_object.ehlo()
```

To initiate encryption, grab the object

```python
smtp_object.starttls()
```

Set up the E-Mail and Passwords

```python
password = input('What is your password: ')
```

Generate an App Password

```python
email = getpass.getpass("Email: ")
password = getpass.getpass("Password: ")
smtp_object.login(email,password)
```
