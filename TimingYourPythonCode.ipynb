Method 1

```python
def func_one(n):
  return [str(num) for num in range(n)]
```

```python
func_one(10)
```

Method 2

```python
def func_two(n):
  return list(map(str,range(n)))
```

```python
func_two(10)
```

Which one is more efficient?

Time Your Code: Mark a start time, calculate the difference

```python
import time
```

```python
# CURRENT TIME BEFORE
start_time = time.time()
# RUN CODE
result = func_one(1000000)
# CURRENT TIME AFTER RUNNING CODE
end_time = time.time()
# ELAPSED TIME
elapsed_time = end_time - start_time

print(elapsed_time)
```

```python
# CURRENT TIME BEFORE
start_time = time.time()
# RUN CODE
result = func_two(1000000)
# CURRENT TIME AFTER RUNNING CODE
end_time = time.time()
# ELAPSED TIME
elapsed_time = end_time - start_time

print(elapsed_time)
```

Use the timeit module that is specifically designed to time code

```python
import timeit
```

```python
timeit.timeit
```

For Function 1:

```python
stmt = '''
func_one(100)
'''
```

```python
setup = '''
def func_one(n):
  return [str(num) for num in range(n)]
'''
```

```python
timeit.timeit(stmt,setup,number=100000)
```

```python
timeit.timeit(stmt,setup,number=1000000)
```

For Function 2:

```python
stmt2 = '''
func_two(100)
'''
```

```python
setup2 = '''
def func_two(n):
  return list(map(str,range(n)))
'''
```

```python
timeit.timeit(stmt2,setup2,number=100000)
```

```python
timeit.timeit(stmt2,setup2,number=1000000)
```
