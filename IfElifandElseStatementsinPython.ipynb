If Statement

```python
if True:
  print('ITS TRUE')
```

Comparison Operation

```python
if 3>2:
  print('ITS TRUE')
```

```python
hungry = True

if hungry:
  print('FEED ME!')
```

```python
hungry = False

if hungry:
  print('FEED ME!')
```

Adding in an Else statement if the condition does not execute

```python
hungry = False

if hungry:
  print('FEED ME!')
else:
  print('Im not hungry')
```

Adding elif

```python
loc = 'Bank'

if loc == 'Auto Shop':
  print("Cars are cool!")
else:
  print("I do not know much.")
```

```python
loc = 'Bank'

if loc == 'Auto Shop':
  print("Cars are cool!")
elif loc == 'Bank':
  print("Money is cool!")
else:
  print("I do not know much.")
```
