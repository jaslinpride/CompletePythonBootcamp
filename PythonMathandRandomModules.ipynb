Why does 4.5 round down while 5.5 round up?

You always choose to round to all evens or all odds. If you always choose to round down when it comes to 0.5 split then all your estimates over time would be lower than they should be. If you always choose to round up when it comes to a 0.5 split then all your estimates over time would be higher than they should be. Choosing a rule based on even or odd evens the math out.

Testing a code script and you want to make adjustments to this script of code but want to test it on random numbers but the same batch

Seed: start from a seeded pseudo-random number generator, meaning the same random numbers show up in a series

```python
random.seed(101)

random.randint(0,100)
```

```python
random.randint(0,100)
```

```python
random.randint(0,100)
```

```python
random.seed(101)
print(random.randint(0,100)) #74
print(random.randint(0,100)) #24
print(random.randint(0,100)) #69
print(random.randint(0,100))
print(random.randint(0,100))
```

Random Library: Taking a random item from a list

```python
mylist = list(range(0,20))
```

```python
mylist
```

```python
random.choice(mylist)
```

The random.choice does not permanently affect your list:

```python
mylist
```

Random Library: Taking multiple random items from a list

Method 1: Allow yourself to pick the same number twice (sampling with replacement)

```python
random.choices(population=mylist,k=10)
```

Method 2: Picking an item from a list and not being able to pick it again (sampling without replacement)

```python
random.sample(population=mylist,k=10)
```

How to Shuffle a List

```python
random.shuffle(mylist)
```
