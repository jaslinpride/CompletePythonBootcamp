```python
'phone' in text
```

```python
import re
```

```python
pattern = 'phone'
```

The match object reports back matches to the phone and where the index location spans to (index 12 - 17)

```python
re.search(pattern,text)
```

Running the Search Again

```python
pattern = 'phone'
```

```python
match = re.search(pattern,text)
```


This is a Regular Expression match object that has information that can be taken from it:

```python
match
```

```python
match.span()
```

```python
match.start()
```

```python
match.end()
```

If we had multiple matches inside the string we would unfortunately only get back the first match

```python
text = 'my phone once, my phone twice'
```

```python
match = re.search('phone',text)
```

```python
match
```

If you want to find multiple matches or all the matches you use the “findall” function

```python
matches = re.findall('phone',text)
```

```python
matches
```

```python
len(matches)
```

To get the match objects use the “iterator” function

```python
for match in re.finditer('phone',text):
  print(match)
```

```python
for match in re.finditer('phone',text):
  print(match.span())
```

Use the “Group” method for the actual text that matches

```python
for match in re.finditer('phone',text):
  print(match.group())
```
