Being able to call methods that affect attributes

```python
class Simple():

  def __init__(self,value):

    self.value = value

#Takes in the current value of self or its current self dot value
  def add_to_value(self,amount):

    self.value = self.value + amount
```

add_to_value does not return anything. It only takes the amount that is passed in, and then adds it to the value.

```python
myobj = Simple(300)
```

```python
myobj.value
```

Nothing is returned because all that is happening when you call this method is it takes in the amount you passed in. It grabs your current value, adds the amount to it, then reassigns self.value to reflect the change.

```python
myobj.add_to_value(500)
```

```python
myobj.value
```

```python
class Simple():

  def __init__(self,value):

    self.value = value

#Takes in the current value of self or its current self dot value
  def add_to_value(self,amount):

    self.value = self.value + amount
    print('We just added {} to your value'.format(amount))
```

```python
myobj = Simple(300)
```

```python
myobj.add_to_value(500)
```

```python
myobj.value
```
