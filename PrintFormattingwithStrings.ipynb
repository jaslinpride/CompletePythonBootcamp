Formatting with the .format() method
A good way to format objects into your strings for print elements is with the string .format() method. The syntax is:

'String here {} then also {}'.format('something1','something2')

```python
print('This is a string {}'.format('INSERTED'))
```

Strings can be inserted by index position

```python
print('The {} {} {}'.format('fox','brown','quick'))
```

```python
print('The {2} {1} {0}'.format('fox','brown','quick'))
```

```python
print('The {0} {0} {0}'.format('fox','brown','quick'))
```

You can also assign key words and call the key words

```python
print('The {q} {b} {f}'.format(f='fox',b='brown',q='quick'))
```

```python
print('The {f} {f} {f}'.format(f='fox',b='brown',q='quick'))
```

Float formatting follows "{value:width.precision *f*}"

```python
result = 100 /777
```

```python
result
```

Can change the level of precision desired

```python
print("The result was {}".format(result))
```

```python
print("The result was {r}".format(r=result))
```

```python
print("The result was {r:1.3f}".format(r=result))
```

Making the width value large (adding white space)

```python
print("The result was {r:10.3f}".format(r=result))
```

```python
print("The result was {r:10.5f}".format(r=result))
```

```python
result = 104.12345
```

```python
print("The result was {r:1.5f}".format(r=result))
```

```python
print("The result was {r:1.2f}".format(r=result))
```

```python
print('Hello, his name is {}'.format(name))
```
