Loop through a list of tuples and unpack the values within them

```python
stock_prices = [('APPL',200),('GOOG',400),('MSFT',100)]
```

```python
for item in stock_prices:
  print(item)
```

Tuple Unpacking: individually grab the items from these tuples

```python
for ticker,price in stock_prices:
  print(ticker)
```

```python
for ticker,price in stock_prices:
  print(price+(0.1*price))
```

Function returning things as tuples and unpacking the result from the function


```python
work_hours = [('Abby',100),('Billy',400),('Cassie',800)]
```

Function that goes through the tuples and sees the name and corresponding hours and keep track of the name that has the most hours.

```python
def employee_check(work_hours):

  current_max = 0
  employee_of_month = ''

  for employee,hours in work_hours:
    if hours > current_max:
      current_max = hours
      employee_of_month = employee
    else:
      pass

  # Return
  return (employee_of_month,current_max)
```

```python
employee_check(work_hours)
```
