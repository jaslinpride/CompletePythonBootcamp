To print only the even numbers in the list

```python
for num in mylist:
  # Check for even by looking for the remainder (Mod operator)
  if num % 2 == 0:
    print(num)
```

```python
for num in mylist:
  # Check for even by looking for the remainder (Mod operator)
  if num % 2 == 0:
    print(num)
  else:
    print('Odd Number')
```

```python
for num in mylist:
  # Check for even by looking for the remainder (Mod operator)
  if num % 2 == 0:
    print(num)
  else:
    print(f'Odd Number: {num}')
```

Indentation Logic: To get the sum of every number in list

```python
list_sum = 0

for num in mylist:
  list_sum = list_sum + num

print(list_sum)
```

Indentation Logic: print statement INSIDE for loop

```python
list_sum = 0

for num in mylist:
  list_sum = list_sum + num

  print(list_sum)
```

If you are iterating through a sequence that contains itself, tuples, the item can be used with tuple unpacking.

You have four items in the list where every single item is a tuple pair

```python
len(mylist)
```

Tuple Unpacking: duplicate the structure of the items (items) in the sequence and unpack them to access individual items

```python
for (a,b) in mylist:
  print(a)
  print(b)
```

```python
for (a,b) in mylist:
  print(a)
```

```python
for (a,b) in mylist:
  print(b)
```

How to Iterate through a Dictionary

```python
d = {'k1':1,'k2':2,'k3':3}

for item in d:
  print(item)
```

How to Iterate through the Items themselves

```python
d = {'k1':1,'k2':2,'k3':3}

for item in d.items():
  print(item)
```

Tuple Unpacking

```python
d = {'k1':1,'k2':2,'k3':3}

for key,value in d.items():
  print(value)
```
