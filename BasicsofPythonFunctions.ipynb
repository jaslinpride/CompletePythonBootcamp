The main point of a function is to execute the block of code, especially when you plan on repeating that block of code.  

```python
def say_hello(name):
  print(f'Hello {name}')
```

```python
say_hello('Jose')
```

```python
def say_hello(jelly):
  print(f'Hello {jelly}')
```

```python
say_hello('Jose')
```

If you forget to pass through a value in “say_hello()” you will get back TypeError: say_hello() missing 1 required positional argument: ‘name’
Either
Provide a value: ‘name’
Default Value: = ‘Default’


```python
def say_hello(name='Default'):
  print(f'Hello {name}')
```

What is the difference between print in a function versus return in a function?
Return: instead of just printing out the result it will actually allow you to save them as a variable


```python
def add_num(num1,num2):
  return num1+num2
```

```python
add_num(10,20)
```

Because you are using return you are able to assign it to a variable

```python
result = add_num(10,20)
```

```python
result
```
