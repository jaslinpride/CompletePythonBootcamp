LESSER OF TWO EVENS: Write a function that returns the lesser of two given numbers if both numbers are even, but returns the greater if one or both numbers are odd

```python
def lesser_of_two_evens(a,b):
  if a%2 == 0 and b%2 == 0:
    # BOTH NUMBERS ARE EVEN!
    if a < b:
      result = a
    else:
      result = b
  else:
    # ONE OR BOTH NUMBERS ARE ODD!
    if a > b:
      result = a
    else:
      result = b

  return result
```

```python
lesser_of_two_evens(2,4)
```

```python
lesser_of_two_evens(2,5)
```

```python
lesser_of_two_evens(7,5)
```

ANIMAL CRACKERS: Write a function that takes a two-word string and returns True if both words begin with the same letter

```python
def animal_crackers(text):
  wordlist = text.split()
  print(wordlist)

  first = wordlist[0]
  second = wordlist[1]

  return first[0] == second[0]
```

MAKES TWENTY: Given two integers, return True if the sum of the integers is 20 or if one of the integers is 20. If not, return False.

```python
def makes_twenty(n1,n2):
  if n1 + n2 == 20:
    return True
  elif n1 == 20:
    return True
  elif n2 == 20:
    return True
  else:
    return False
```

```python
makes_twenty(20,10)
```

```python
makes_twenty(2,3)
```

```python
makes_twenty(10,10)
```

Consolidate to a single line using comparison operators and logical operators like "Or"

```python
def makes_twenty(n1,n2):
  return (n1+n2) == 20 or n1 == 20 or n2 == 20
```
