Class Object Attribute: There may be attributes that are going to be the same for any instance of the dog class, e.g., it does not matter whether a dog has a certain breed, a certain name, whether or not it has spots, it is always going to be a mammal. Therefore, we may want an attribute be defined at a class object level instead of having the particular instance. We will define an attribute above our __init__ method.

Do not use the self keyword because it is a reference to this particular instance of a class.  


```python
class Dog():

    # CLASS OBJECT ATTRIBUTE
    # SAME FOR ANY INSTANCE OF A CLASS
    species = 'mammal'

    def __init__(self, breed, name, spots):

      # ATTRIBUTES
      # WE TAKE IN THE ARGUMENT
      # ASSIGN IT USING self.attribute_name
      self.breed = breed
      self.name = name

      # EXPECT boolean True/False
      self.spots = spots
```

```python
my_dog = Dog(breed='Lab',name='Sam',spots=False)
```

```python
type(my_dog)
```

```python
my_dog.species
```

```python
class Dog():

    # CLASS OBJECT ATTRIBUTE
    # SAME FOR ANY INSTANCE OF A CLASS
    species = 'mammal'

    def __init__(self, breed, name):

      # ATTRIBUTES
      # WE TAKE IN THE ARGUMENT
      # ASSIGN IT USING self.attribute_name
      self.breed = breed
      self.name = name

    # OPERATIONS/ACTIONS ---> Methods
    def bark(self):
      print("WOOF!")
```

```python
class Circle():

  # CLASS OBJECT ATTRIBUTE
  pi = 3.14

  def __init__(self,radius=1):

    self.radius = radius

  # METHOD
  def get_circumference(self):
    return self.radius * self.pi * 2
```

```python
my_circle = Circle()
```

```python
my_circle.pi
```

Radius = 1 due to the default value provided above

```python
my_circle.radius
```

```python
my_circle = Circle(30)
```

```python
my_circle.pi
```

```python
my_circle.radius
```

```python
my_circle.get_circumference()
```

If you have an attribute it does not necessarily have to be defined from a particular parameter call

```python
class Circle():

  # CLASS OBJECT ATTRIBUTE
  pi = 3.14

  def __init__(self,radius=1):

    self.radius = radius
    self.area = radius*radius*self.pi

  # METHOD
  def get_circumference(self):
    return self.radius * self.pi * 2
```

```python
my_circle = Circle(30)
```

```python
my_circle.pi
```

```python
my_circle.radius
```

```python
my_circle.area
```
