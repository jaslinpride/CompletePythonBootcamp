```python
def myfunc(a,b):
  # Returns 5% of the sum of a and b
  return sum((a,b)) * 0.05
```

```python
myfunc(40,60)
```

What if I had more numbers or parameters to pass in?


```python
def myfunc(a,b,c=0,d=0,e=0):
  # Returns 5% of the sum of a and b
  return sum((a,b,c,d,e)) * 0.05
```

```python
myfunc(40,60,100)
```

```python
myfunc(40,60,100,100)
```

*args allows you to take an arbitrary number of arguments. It treats it as a tuple of parameters that are coming in.


```python
def myfunc(*args):
  return sum(args) * 0.05
```

```python
myfunc(40,60,100,1,34)
```

Just printing args

```python
def myfunc(*args):
  print (args)
```

```python
myfunc(40,60,100,1,34)
```

Args are denoted by the * symbol and can be any keyword you choose

```python
def myfunc(*spam):
  print(spam)
```

```python
myfunc(40,60,100,1,34)
```

As a Tuple

```python
def myfunc(*args):
  for item in args:
    print(item)
```

```python
myfunc(40,60,100,1,34)
```

Building a Dictionary of Key Value Pairs

```python
def myfunc(**kwargs):
  if 'fruit' in kwargs:
    print('My fruit of choice is {}'.format(kwargs['fruit']))
  else:
      print('I did not find any fruit here')
```

```python
myfunc(fruit='apple')
```

```python
def myfunc(**kwargs):
  print(kwargs)
  if 'fruit' in kwargs:
    print('My fruit of choice is {}'.format(kwargs['fruit']))
  else:
      print('I did not find any fruit here')
```

```python
myfunc(fruit='apple',veggie = 'lettuce')
```

kwargs are denoted by the * symbol and can be any keyword you choose. By convention, always use “kwargs”.

```python
def myfunc(**jelly):
  print(jelly)
```

```python
myfunc(fruit='apple',veggie = 'lettuce')
```

*args and **kwargs can be used in combination

```python
def myfunc(*args,**kwargs):

  print('I would like {} {}'.format(args[0],kwargs['food']))
```

```python
myfunc(10,20,30,fruit='orange',food='eggs',animal='dog')
```

```python
def myfunc(*args,**kwargs):
  print(args)
  print(kwargs)
  print('I would like {} {}'.format(args[0],kwargs['food']))
```

```python
myfunc(10,20,30,fruit='orange',food='eggs',animal='dog')
```

Because we said arguments first and keyword arguments second it has to go in that order in the code.
