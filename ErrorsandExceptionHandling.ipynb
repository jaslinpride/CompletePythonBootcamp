Finally Keyword will have a block of code that is always executing

```python
try:
  f = open('testfile','w')
  f.write("Write a test line")
except TypeError:
  print("There was a type error!")
except OSError:
  print('Hey you have an OS Error')
finally:
  print("I always run")
```

How to use Try, Except, Finally blocks inside of a function that tries to get a specific type of input from a user

```python
def ask_for_int():
  try:
    result = int(input("Please provide number: "))
  except:
    print("Whoops! That is not a number")
  finally:
    print("End of try/except/finally")
```

```python
ask_for_int()
```

```python
ask_for_int()
```

Using a while loop to continually be doing this try block of code

```python
def ask_for_int():
  while True:
    try:
      result = int(input("Please provide number: "))
    except:
      print("Whoops! That is not a number")
      continue
    else:
      print("Yes thank you")
      break
    finally:
      print("End of try/except/finally")
      print("I will always run at the end!")
```

```python
ask_for_int()
```

```python
ask_for_int()
```
