Now define a function inside a function

You have the function “hello” it takes in some default name, “Jose”, and prints out “The hello() function has been executed!”. Inside of this function, you are defining another function called “greet” and it returns “This is the greet() func inside hello!”. You are just defining “greet” you are not actually calling it.  

```python
def hello(name='Jose'):
  print('The hello() function has been executed!')

  def greet():
    return '\t This is the greet() func inside hello!'
```

```python
hello()
```

Calling “greet”

```python
def hello(name='Jose'):
  print('The hello() function has been executed!')

  def greet():
    return '\t This is the greet() func inside hello!'

  print(greet())
```

```python
hello()
```

Defining another function inside “hello”

```python
def hello(name='Jose'):
  print('The hello() function has been executed!')

  def greet():
    return '\t This is the greet() func inside hello!'

  def welcome():
    return '\t This is the welcome() inside hello!'

  print(greet())
  print(welcome())
  print('This is the end of the hello function!')
```

```python
hello()
```

What if you want to access these functions outside of hello?

```python
def hello(name='Jose'):
  print('The hello() function has been executed!')

  def greet():
    return '\t This is the greet() func inside hello!'

  def welcome():
    return '\t This is the welcome() inside hello!'

  print("I am going to return a function!")

  if name == 'Jose':
    return greet
  else:
    return welcome
```

```python
my_new_func = hello('Jose')
```

```python
my_new_func
```

```python
my_new_func()
```

```python
print(my_new_func())
```

You can return functions and you have functions as arguments. Now, you can create a decorator.

```python
def new_decorator(original_func):

  def wrap_func():

    print('Some extra code, before the original function')

    original_func()

    print('Some extra code, after the original function')

  return wrap_func
```

```python
def func_needs_decorator():
  print("I want to be decorated!!")
```

```python
func_needs_decorator()
```

Add in some extra code using the framework you built in In [39]:

```python
decorated_func = new_decorator(func_needs_decorator)
```

```python
decorated_func()
```

There is a special syntax for “decorated_func = new_decorator(func_needs_decorator)”. The @ operator

```python
@new_decorator
def func_needs_decorator():
  print("I want to be decorated!!")
```

```python
func_needs_decorator()
```

To turn decorators off just comment out “@new_decorator” using hashtags

```python
# @new_decorator
def func_needs_decorator():
  print("I want to be decorated!!")
```

```python
func_needs_decorator()
```
