```python
d = {'k1':1,'k2':2}
```

Creating a Dictionary through Dictionary Comprehension

```python
{x:x**2 for x in range(10)}
```

Assigning Keys that are not Based on the Values

```python
{k:v**2 for k,v in zip(['a','b'],range(2))}
```

Iteration over Keys, Values, Items

```python
for k in d.iteritems():
  print(k)
```

```python
for k in d.itervalues():
  print(k)
```

```python
for k in d.iterkeys():
  print(k)
```

View Method

```python
d.viewitems
```

```python
d.viewitems()
```

```python
d.viewkeys()
```

```python
d.viewvalues()
```
