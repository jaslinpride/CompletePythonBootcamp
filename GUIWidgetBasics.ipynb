Inserting a number into the text box does not change the position of the slider

```python
a = widgets.FloatText()
b = widgets.FloatSlider()
display(a,b)
```

```python
display(a)
```

Linking the number in the text box and the position of the slider

```python
a = widgets.FloatText()
b = widgets.FloatSlider()
display(a,b)

mylink = widgets.jslink((a,'value'),(b,'value'))
```

Linking “max”

```python
a = widgets.FloatText()
b = widgets.FloatSlider()
display(a,b)

mylink = widgets.jslink((a,'value'),(b,'max'))
```

Unlinking Keys

```python
mylink.unlink()
```
