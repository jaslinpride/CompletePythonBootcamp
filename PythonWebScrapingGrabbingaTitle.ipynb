Grabbing a Page Title from Any Website

```python
import requests
```

Get a Web Page through a Request

```python
result = requests.get("http://www.example.com")
```

```python
type(result)
```

```python
result.text
```

Convert the String into a Soup using BeautifulSoup

```python
import bs4
```

Create the Soup Variable

```python
soup = bs4.BeautifulSoup(result.text,"lxml")
```

```python
soup
```

Grab the Title, “Example Domain” from the soup

```python
soup.select('title')
```

```python
soup.select('p')
```

```python
soup.select('h1')
```

Grabbing the First Item from the Title

```python
soup.select('title')[0]
```

Get Text Method

```python
soup.select('title')[0].getText()
```

Grabbing a String from the Text

```python
site_paragraphs = soup.select("p")
```
