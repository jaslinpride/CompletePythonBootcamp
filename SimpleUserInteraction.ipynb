```python
game_list = [0,1,2]
from IPython.display import clear_output
clear_output()
```

Have a Function that can Display the Game

```python
def display_game(game_list):
  print("Here is the current list")
  print(game_list)
```

```python
display_game(game_list)
```

Have a Function where the User Chooses a Position

```python
def position_choice():

  #This original choice value can be anything that is not an integer
  choice = 'wrong'

  #While the choice is not a digit, keep asking for input.
  while choice not in ['0','1','2']:

    # We should not convert here, otherwise we get an error on a wrong input
    choice = input("Pick a position to replace (0,1,2): ")

    if choice not in ['0','1','2']:
      # THIS CLEARS THE CURRENT OUTPUT BELOW THE CELL
      clear_output()

      print("Sorry, but you did not choose a valid position (0,1,2)")

  # Optionally you can clear everything after running the function
  # clear_output()

  # We can convert once the while loop above has confirmed we have a digit.
  return int(choice)
```

Testing the section above:

```python
position_choice()
```

Replacement Choice Function

```python
def replacement_choice(game_list,position):

  user_placement = input("Type a string to place at the position: ")

  game_list[position] = user_placement

  return game_list
```

Testing the Section Above:

```python
replacement_choice(game_list,1)
```

Choice asking user if they want to keep playing

```python
def gameon_choice():

  #This original choice value can be anything that is not a Y or N
  choice = 'wrong'

  # While the choice is not a digit, keep asking for input.
  while choice not in ['Y','N']:

    # We should not convert here, otherwise we get an error on a wrong input
    choice = input("Would like to keep playing? Y or N ")

    if choice not in ['Y','N']:
      # THIS CLEARS THE CURRENT OUTPUT BELOW THE CELL
      clear_output()

      print("Sorry, I did not understand. Please make sure to choose Y or N.")

  # Optionally you can clear everything after running the function
  # clear_output()

  if choice == "Y":
    # Game is still on
    return True
  else:
    # Game is over
    return False
```

Testing the Section Above:

```python
gameon_choice()
```

```python
# Variable to keep game playing
game_on = True

# First Game List
game_list = [0,1,2]



while game_on:

  # Clear any historical output and show the game list
  clear_output()
  display_game(game_list)

  # Have player choose position
  position = position_choice()

  # Rewrite that position and update game list
  game_list = replacement_choice(game_list,position)

  # Clear Screen and show the updated game list
  clear_output()
  display_game(game_list)

  # Ask if you want to keep playing
  game_on = gameon_choice()
```
