Create a Normal Function

```python
def create_cubes(n):
  result = []
  for x in range(n):
    result.append(x**3)
  return result
```

```python
create_cubes(10)
```

When working with a normal function, you have to create an empty list, and then go for every number from 0 up to that value.
You only need one value at a time to print you do not need a whole list stored in memory. You just need the previous value and whatever formula is to get to the next value in order to generate all the cubes.


```python
for x in create_cubes(10):
  print(x)
```

It is better to yield the actual cube numbers

```python
def create_cubes(n):

  for x in range(n):
    yield x**3
```

```python
for x in create_cubes(10):
  print(x)
```

Calling “create_cubes” by itself
Iterate through it for the actual list of numbers


```python
create_cubes(10)
```

```python
list(create_cubes(10))
```

Calculating a Fibonacci Sequence

Fibonacci Sequence: If you have a number A followed by a number B, then B is defined as the sum of the last two numbers


```python
def gen_fibon(n):

  a = 1
  b = 1
  for i in range(n):
    yield a
    a,b = b,a+b
```

```python
for number in gen_fibon(10):
  print(number)
```

The difference between “yielding” and creating a normal function


```python
def gen_fibon(n):

  a = 1
  b = 1
  output = []

  for i in range(n):
    output.append(a)
    a,b = b,a+b
  return output
```

The key to fully understanding generators is the “next” function and the “iter” function

Next Function

```python
def simple_gen():
  for x in range(3):
    yield x
```

```python
for number in simple_gen():
  print(number)
```

```python
g = simple_gen()
```

```python
g
```

```python
print(next(g))
```

```python
print(next(g))
```

Internally, the generator is remembering what the previous answer was and then returning the next value given whatever formula it is following. It is not holding everything in memory.
