Avoid using “global” keyword unless absolutely necessary. Instead, grab the global variable by taking it in as a parameter, do the reassignment, and then return the assignment as an object itself.

```python
x = 50

def func(x):

  print(f'X is {x}')

  # LOCAL REASSIGNMENT ON A GLOBAL VARIABLE!
  x = 'NEW VALUE'
  print(f'I JUST LOCALLY CHANGED X TO {x}')
  return x
```

```python
print(x)
```

```python
x = func(x)
```
