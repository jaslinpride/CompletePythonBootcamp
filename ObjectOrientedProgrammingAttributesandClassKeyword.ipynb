Changing the Name for Clarity: When you create an instance of the dog class, Python is going to call the __init__ method, and it will use “self” in order to represent the instance of the object itself. It expects you to pass in the argument “mybreed”

```python
class Dog():

  def __init__(self, mybreed):

    # ATTRIBUTES
    # We take in the argument
    # Assign it using self.attribute_name
    self.breed = mybreed
```

```python
my_dog = Dog(mybreed='Huskie')
```

```python
type(my_dog)
```

```python
my_dog.breed
```

```python
class Dog():

  def __init__(self, mybreed):

    # ATTRIBUTES
    # We take in the argument
    # Assign it using self.attribute_name
    self.my_attribute = mybreed
```

```python
my_dog = Dog(mybreed='Huskie')
```

```python
type(my_dog)
```

```python
my_dog.my_attribute
```

Giving the dog more attributes:

```python
class Dog():

  def __init__(self, breed,name,spots):

    # ATTRIBUTES
    # We take in the argument
    # Assign it using self.attribute_name
    self.breed = breed
    self.name = name

    # Expect boolean True/False
    self.spots = spots
```

```python
my_dog = Dog(breed='Lab',name='Sammy',spots=False)
```

```python
type(my_dog)
```

```python
my_dog.breed
```

```python
my_dog.name
```

```python
my_dog.spots
```

```python
my_dog = Dog(breed='Lab',name='Sammy',spots="NO SPOTS")
```

```python
my_dog.spots
```
