**Problem 1: Handle the exception thrown by the code below by using try and except blocks.**

```python
for i in ['a','b','c']:
  print(i**2)
```

Put the code in a “try” block

```python
try:
  for i in ['a','b','c']:
    print(i**2)
except TypeError:
  print("Type error! Watch out!")
```

**Problem 2: Handle the exception thrown by the code below by using try and except blocks. Then use a finally block to print ‘All Done.’ **

```python
x = 5
y = 0

z = x/y
```

Put the code in a “try” block

```python
try:
  x = 5
  y = 0
  z = x/y
except:
  print("Error!!")
finally:
  print("All done")
```

“while True” is dependent on the “break” keyword.
Instead, you can


```python
def ask():

  # Waiting for correct response
  waiting = True
  while waiting:
    try:
      n = int(input("Enter a number"))
    except:
      print("Please try again! \n")
      continue
    else:
      waiting = False

  print("Your number squared is: ")
  print(n**2)
```
