```python
d
```

Dictionary Methods: Reassign

```python
d = {'k1': 100, 'k2': 200, 'k3': 300}
```

To see all the keys of a dictionary

```python
d.keys()
```

To see all the values of a dictionary

```python
d.values()
```

To see all the pairings of a dictionary

```python
d.items()
```
