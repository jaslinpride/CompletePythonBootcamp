Color Transparency

RGBA: Red, Green, Blue, Alpha

All RGBA values can go from 0 to 255
Completely Transparent: If the Alpha Value is 0
Completely Opaque: If the Alpha Value is 0


```python
red = Image.open('red_color.jpg')
```

```python
blue = Image.open('blue_color.png')
```

Reset Alpha

```python
blue.putalpha(0)
```

```python
blue
```

```python
blue.putalpha(255)
```

```python
blue
```

```python
blue.putalpha(128)
```

```python
blue
```

```python
red.putalpha(128)
```

```python
red
```

Paste Functionality
