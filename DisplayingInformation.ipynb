Printing Variables

```python
print([1,2,3])
```

Printing multiple lines

```python
print([1,2,3])
print([4,5,6])
print([7,8,9])
```

Instead of calling each print function over and over use a custom function

```python
def display(row1,row2,row3):
  print(row1)
  print(row2)
  print(row3)
```

Display the function you just defined

```python
example_row = [1,2,3]

display(example_row,example_row,example_row)
```

Can be a list, string, or object

```python
row1 = [' ',' ',' ']
row2 = [' ',' ',' ']
row3 = [' ',' ',' ']
```

```python
display(row1,row2,row3)
```

Rudimentary Tic Tac Toe Board

```python
row2[1] = 'X'
```

```python
display(row1,row2,row3)
```
