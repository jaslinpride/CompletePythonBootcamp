Create a function that can take in a player guess


```python
def player_guess():

  guess=''

  while guess not in ['0','1','2']:
    guess = input("Pick a number: 0,1, or 2")

  return int(guess)
```

```python
player_guess()
```

Save player guess

```python
myindex = player_guess()
```

Combine list that has been shuffled with the player's guess

```python
def check_guess(mylist,guess):
  if mylist[guess] == 'O':
    print("Correct!")
  else:
    print("Wrong guess!")
    print(mylist)
```

```python
# INITIAL LIST
mylist = ['','O','']

# SHUFFLE LIST
mixedup_list = shuffle_list(mylist)

# USER GUESS
guess = player_guess()

#CHECK GUESS
check_guess(mixedup_list,guess)
```
