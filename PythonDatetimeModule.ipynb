Performing Arithmetic with Datetime

Example: User Log in Times/Dates (How long did they spend on my website/logged in?)


```python
from datetime import date
```

```python
date1 = date(2021,11,3)
date2 = date(2020,11,3)
```

```python
date1 - date2
```

```python
result = date1 - date2
```

```python
type(result)
```

```python
result.days
```

If you are doing arithmetic with really long distances you may have issues with leap years.

```python
datetime2 = datetime(2020,11,3,12,0)
```

```python
datetime1 - datetime2
```
